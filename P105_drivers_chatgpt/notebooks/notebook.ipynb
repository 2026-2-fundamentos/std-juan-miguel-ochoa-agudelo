# Análisis de conductores con ChatGPT

Este notebook reutiliza `summary.csv` y `top10_drivers.png`, generados en P104 y copiados a `submission/`. El resumen contiene las horas y millas acumuladas por conductor. En esta exploración ordenamos a los conductores por millas registradas y revisamos el gráfico.

Ejecuta el notebook desde la carpeta `P105_drivers_chatgpt`.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

submission = Path("submission")
drivers = pd.read_csv(submission / "summary.csv")
top10 = drivers.nlargest(10, "miles_logged")
display(top10[["driver_id", "name", "hours_logged", "miles_logged"]])
display(Image(filename=str(submission / "top10_drivers.png")))
```
